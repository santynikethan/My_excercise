####How to run a Spark SQL Queries inside a notebook using %sql magic command

```python
%sql
create table if not exists workspace.default.players(id int, name string);
insert into default.players values(1,'Kholi'),(2,'Rohit'),(3,'Dhoni'),(4,'Gambhi');
select count(*) from players;
```

```python
#If we want to perform both Func based prog & Declartive prog.
spark.sql("select * from players").count()
```

```python
%sql
update players set name = 'Virat Kholi' where id = 1;
update players set name = 'Rohit Sharma' where id = 2;
update players set name = 'MS Dhoni' where id = 3;
update players set name = 'Gautam Ghambir' where id = 4;
select * from players;
```
